```python
import os
from PIL import Image
from pillow_heif import register_heif_opener

# Register HEIF opener with Pillow
register_heif_opener()

def convert_heic_to_jpg(source_folder, quality=90):
    # Create a destination folder
    destination_folder = os.path.join(source_folder, "converted_jpgs")
    os.makedirs(destination_folder, exist_ok=True)

    for filename in os.listdir(source_folder):
        if filename.lower().endswith(".heic"):
            heic_path = os.path.join(source_folder, filename)
            
            # Open image
            image = Image.open(heic_path)
            
            # Create new filename with .jpg extension
            new_filename = os.path.splitext(filename)[0] + ".jpg"
            jpg_path = os.path.join(destination_folder, new_filename)
            
            # Save as JPG
            # "convert('RGB')" is needed because some HEICs are RGBA
            image.convert("RGB").save(jpg_path, "JPEG", quality=quality)
            print(f"Converted: {filename}")

# Usage
folder_path = r"C:\Users\tlhen\Downloads\ball"
convert_heic_to_jpg(folder_path)
```
