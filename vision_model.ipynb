```python
from ultralytics import YOLO
import cv2

# Load YOUR custom model
model = YOLO('best.pt') 

# Initialize Webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret: break
    
    # Run inference
    # conf=0.5 means "only show me things you are 50% sure are balls"
    results = model(frame, conf=0.5)
    
    # Visualize the results
    annotated_frame = results[0].plot()
    
    cv2.imshow("My Custom AI", annotated_frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
        
cap.release()
cv2.destroyAllWindows()
```
